```python
import numpy as np
import cv2

def align_images(img_a, img_b, kps_a, kps_b, matches,
                 method='homography', ransac_thresh=5.0, min_matches=4):
    """
    Align img_b to img_a using matched keypoints.

    Parameters:
    - img_a, img_b: np.ndarray images (H,W[,C]) where img_b will be warped into img_a frame.
    - kps_a, kps_b: arrays of keypoint coordinates. Each can be shape (N,2) containing (x,y).
    - matches: either
        * list/array of (i,j) index pairs (index into kps_a and kps_b),
        * or boolean/integer mapping array such that for each i in kps_a there is matched j,
        * or None if kps_a and kps_b are already pairwise and same length.
    - method: 'homography' or 'affine'
    - ransac_thresh: RANSAC reprojection threshold (pixels)
    - min_matches: minimum matches required to run estimation

    Returns:
    - warped_b: img_b warped into img_a coordinate frame (same dtype as input)
    - M: transformation matrix (3x3 for homography, 2x3 for affine)
    - mask: inlier mask from RANSAC (array of 0/1)
    """
    # Build matched point arrays
    if matches is None:
        pts_a = np.asarray(kps_a, dtype=np.float32)
        pts_b = np.asarray(kps_b, dtype=np.float32)
    else:
        m = np.asarray(matches)
        if m.ndim == 2 and m.shape[1] >= 2:
            idx_a = m[:, 0].astype(int)
            idx_b = m[:, 1].astype(int)
        if m.ndim == 1:                         # matches array mapping a->b or -1
            valid = (m != -1)
            idx_a = np.nonzero(valid)[0]
            idx_b = m[valid].astype(int)
        else:
            raise ValueError("matches must be Nx2 indices or None")
        pts_a = np.asarray(kps_a, dtype=np.float32)[idx_a]
        pts_b = np.asarray(kps_b, dtype=np.float32)[idx_b]

    if pts_a.shape[0] < min_matches:
        raise RuntimeError(f"Not enough matches: {pts_a.shape[0]} (min {min_matches})")

    if method == 'homography':
        M, mask = cv2.findHomography(pts_b, pts_a, cv2.RANSAC, ransac_thresh)
        if M is None:
            raise RuntimeError("Homography estimation failed")
        h_a, w_a = img_a.shape[:2]
        warped_b = cv2.warpPerspective(img_b, M, (w_a, h_a), flags=cv2.INTER_LINEAR)
    elif method == 'affine':
        M, mask = cv2.estimateAffinePartial2D(pts_b, pts_a, method=cv2.RANSAC,
                                              ransacReprojThreshold=ransac_thresh)
        if M is None:
            raise RuntimeError("Affine estimation failed")
        h_a, w_a = img_a.shape[:2]
        warped_b = cv2.warpAffine(img_b, M, (w_a, h_a), flags=cv2.INTER_LINEAR)
    else:
        raise ValueError("method must be 'homography' or 'affine'")

    mask = mask.ravel() if mask is not None else None
    return warped_b, M, mask
```

```python
import numpy as np
import cv2

def similarity_transform_from_matches(pts_a, pts_b, eps=1e-9):
    """
    pts_a, pts_b: (N,2) arrays of matched coordinates (A target, B source)
    returns: 2x3 affine matrix M mapping B -> A (apply with cv2.warpAffine)
    """
    A = np.asarray(pts_a, dtype=np.float64)
    B = np.asarray(pts_b, dtype=np.float64)
    if A.shape != B.shape or A.shape[0] < 2:
        raise ValueError("pts_a and pts_b must be same shape and have >=2 points")

    mu_A = A.mean(axis=0)
    mu_B = B.mean(axis=0)
    A_c = A - mu_A
    B_c = B - mu_B

    H = B_c.T @ A_c  # 2x2
    U, S, Vt = np.linalg.svd(H)
    R = U @ Vt
    if np.linalg.det(R) < 0:
        Vt[1,:] *= -1
        R = U @ Vt

    var_B = (B_c**2).sum()
    if var_B < eps:
        raise RuntimeError("Degenerate B points (zero variance)")

    s = S.sum() / var_B
    t = mu_A - s * (R @ mu_B)

    M = np.hstack((s * R, t.reshape(2,1)))   # 2x3
    return M

def warp_image_with_similarity(img_src, M, out_shape):
    """
    img_src: source image (to be mapped by M to target frame)
    M: 2x3 matrix (from similarity_transform_from_matches)
    out_shape: (width, height) of output (target) canvas
    """
    w, h = out_shape
    return cv2.warpAffine(img_src, M, (w, h), flags=cv2.INTER_LINEAR)

# Helper to build matched point arrays from your 1D mapping (a->b with -1)
def build_pairwise_points_from_mapping(mapping, kps_a, kps_b):
    m = np.asarray(mapping)
    valid = (m != -1)
    idx_a = np.nonzero(valid)[0]
    idx_b = m[valid].astype(int)
    pts_a = np.asarray(kps_a, dtype=np.float64)[idx_a]
    pts_b = np.asarray(kps_b, dtype=np.float64)[idx_b]
    return pts_a, pts_b
```

```python
import numpy as np

def canonical_normalize_keypoints(kps, use_orientation=True, eps=1e-9):
    """
    Normalize 2D keypoints into a canonical frame:
      - translate centroid to origin
      - scale so RMS distance from origin is 1
      - optionally rotate so principal axis aligns with x-axis

    Parameters
    ----------
    kps : (N,2) array
        Original keypoints in pixel coordinates.
    use_orientation : bool
        If True, rotate so the dominant principal component is aligned to x-axis.
    eps : float
        Small value to guard against degenerate configurations.

    Returns
    -------
    kps_norm : (N,2) array
        Keypoints in canonical coordinates.
    M : (2,3) array
        Affine transform mapping original -> canonical coordinates.
        Can be used with cv2.warpAffine on the image.
    Minv : (2,3) array
        Inverse transform mapping canonical -> original coordinates.
    """
    P = np.asarray(kps, dtype=np.float64)
    if P.ndim != 2 or P.shape[1] != 2 or P.shape[0] < 2:
        raise ValueError("kps must be (N,2) with N >= 2")

    # 1) Translate so centroid is at origin
    mu = P.mean(axis=0)
    P_c = P - mu

    # 2) Scale so RMS distance is 1
    sq_norms = (P_c ** 2).sum(axis=1)
    rms = np.sqrt(sq_norms.mean())
    if rms < eps:
        raise RuntimeError("Degenerate keypoints: zero spread")
    s = 1.0 / rms
    P_cs = P_c * s

    # 3) Optional orientation normalization via PCA
    if use_orientation:
        C = np.cov(P_cs.T)    # 2x2 covariance
        eigvals, eigvecs = np.linalg.eigh(C)
        # eigenvectors columns; take the one with largest eigenvalue
        idx_max = np.argmax(eigvals)
        v = eigvecs[:, idx_max]          # principal direction in canonical frame
        # angle of principal axis
        theta = np.arctan2(v[1], v[0])
        # rotation that sends v to +x axis: R * v = [1,0]
        c, sn = np.cos(-theta), np.sin(-theta)
        R = np.array([[c, -sn],
                      [sn,  c]])
    else:
        R = np.eye(2)

    # Apply final rotation
    kps_norm = (R @ P_cs.T).T

    # Build 2x3 affine transforms in homogeneous form
    # canonical = R * s * (x - mu)
    A = R * s                                  # 2x2
    t = -A @ mu                                # 2,
    M = np.hstack([A, t.reshape(2, 1)])        # 2x3

    # Inverse transform: x = R^T * (1/s) * canonical + mu
    Ainv = (R.T) / s
    tinv = mu - Ainv @ np.zeros(2)            # offset from origin canonical (0,0)
    Minv = np.hstack([Ainv, tinv.reshape(2, 1)])

    return kps_norm, M, Minv
```

```python
import numpy as np
import cv2

def warp_image_to_canonical(img, M, scale=100.0, margin=20):
    """
    Warp an image into its canonical keypoint frame.

    img   : source image
    M     : 2x3 transform from canonical_normalize_keypoints (orig -> canonical)
    scale : pixels per canonical unit
    margin: extra pixels around the content
    """
    h, w = img.shape[:2]

    # Original image corners
    corners = np.array([[0, 0],
                        [w, 0],
                        [w, h],
                        [0, h]], dtype=np.float64)

    # Map corners to canonical coordinates
    corners_h = np.hstack([corners, np.ones((4, 1), dtype=np.float64)])  # 4x3
    canon_corners = (M @ corners_h.T).T                                   # 4x2

    # Bounding box in canonical space
    min_xy = canon_corners.min(axis=0)
    max_xy = canon_corners.max(axis=0)

    size_xy = (max_xy - min_xy) * scale
    out_w = int(np.ceil(size_xy[0] + 2 * margin))
    out_h = int(np.ceil(size_xy[1] + 2 * margin))

    # Scale + translate canonical → pixel grid
    tx = -min_xy[0] * scale + margin
    ty = -min_xy[1] * scale + margin
    S = np.array([[scale, 0.0, tx],
                  [0.0, scale, ty]], dtype=np.float64)

    # Combine: orig → canonical → pixel
    M_h = np.vstack([M, [0.0, 0.0, 1.0]])          # 3x3
    M_full = (S @ M_h)[:2, :].astype(np.float32)   # 2x3 for warpAffine

    warped = cv2.warpAffine(img, M_full, (out_w, out_h), flags=cv2.INTER_LINEAR)
    return warped, M_full, (out_w, out_h)

# You already built matched keypoints and canonical transforms:
# pts_a, pts_b = build_pairwise_points_from_mapping(...)
# kps_a_norm, M_a, Minv_a = canonical_normalize_keypoints(pts_a, use_orientation=True)
# kps_b_norm, M_b, Minv_b = canonical_normalize_keypoints(pts_b, use_orientation=True)

canon_img_a, M_a_img, size_a = warp_image_to_canonical(image_a, M_a, scale=100.0, margin=20)
canon_img_b, M_b_img, size_b = warp_image_to_canonical(image_b, M_b, scale=100.0, margin=20)

cv2.imwrite("image_a_canonical.jpg", canon_img_a)
cv2.imwrite("image_b_canonical.jpg", canon_img_b)
```

```python
import numpy as np
import cv2

def build_pairwise(mapping, kps_a, kps_b):
    m = np.asarray(mapping)
    valid = (m != -1)
    idx_a = np.nonzero(valid)[0]
    idx_b = m[valid].astype(int)
    return np.asarray(kps_a)[idx_a].astype(np.float64), np.asarray(kps_b)[idx_b].astype(np.float64), idx_a, idx_b

def ransac_inliers_affine(pts_a, pts_b, thresh=3.0):
    # estimates B->A affine and returns boolean mask of inliers
    M, mask = cv2.estimateAffinePartial2D(pts_b.astype(np.float32), pts_a.astype(np.float32),
                                          method=cv2.RANSAC, ransacReprojThreshold=thresh)
    if mask is None:
        return np.ones(len(pts_a), dtype=bool)
    return mask.ravel().astype(bool)

def canonical_normalize(kps, use_orientation=True, scale_method='rms', eps=1e-9):
    P = np.asarray(kps, dtype=np.float64)
    mu = P.mean(axis=0)
    P_c = P - mu
    d = np.sqrt((P_c**2).sum(axis=1))
    if scale_method == 'median':
        scale_den = np.median(d)
    else:  # 'rms'
        scale_den = np.sqrt((d**2).mean())
    if scale_den < eps:
        raise RuntimeError("Degenerate points")
    s = 1.0 / scale_den
    P_cs = P_c * s
    if use_orientation:
        C = np.cov(P_cs.T)
        eigvals, eigvecs = np.linalg.eigh(C)
        v = eigvecs[:, np.argmax(eigvals)]
        theta = np.arctan2(v[1], v[0])
        c, sn = np.cos(-theta), np.sin(-theta)
        R = np.array([[c, -sn],[sn, c]])
    else:
        R = np.eye(2)
    kps_norm = (R @ P_cs.T).T
    A = R * s
    t = -A @ mu
    M = np.hstack([A, t.reshape(2,1)])    # orig -> canonical
    Minv = np.hstack([(R.T)/s, (mu).reshape(2,1)])  # approx inverse (canonical->orig)
    return kps_norm, M, Minv

def normalize_pairs_consistent(mapping, kps_a, kps_b,
                               ransac_thresh=3.0, use_orientation=True, scale_method='rms'):
    pts_a, pts_b, idx_a, idx_b = build_pairwise(mapping, kps_a, kps_b)
    inliers = ransac_inliers_affine(pts_a, pts_b, thresh=ransac_thresh)
    pts_a_in = pts_a[inliers]
    pts_b_in = pts_b[inliers]
    kpa_norm, Ma, Minv_a = canonical_normalize(pts_a_in, use_orientation=use_orientation, scale_method=scale_method)
    kpb_norm, Mb, Minv_b = canonical_normalize(pts_b_in, use_orientation=use_orientation, scale_method=scale_method)
    return dict(
        pts_a=pts_a, pts_b=pts_b, inliers=inliers,
        pts_a_in=pts_a_in, pts_b_in=pts_b_in,
        kpa_norm=kpa_norm, kpb_norm=kpb_norm,
        M_a=Ma, M_b=Mb, Minv_a=Minv_a, Minv_b=Minv_b
    )

def pairwise_distance_error(kpa_norm, kpb_norm):
    # compare pairwise distance matrices for matched sets (scale already normalized)
    Da = np.sqrt(((kpa_norm[:,None,:]-kpa_norm[None,:,:])**2).sum(-1))
    Db = np.sqrt(((kpb_norm[:,None,:]-kpb_norm[None,:,:])**2).sum(-1))
    # relative RMS error
    num = np.linalg.norm(Da - Db)
    den = np.linalg.norm(Da) + 1e-9
    return num/den

pts = data['matches']         # 1D mapping
kps_a = data['keypoints0']
kps_b = data['keypoints1']

res = normalize_pairs_consistent(pts, kps_a, kps_b, ransac_thresh=50.0, scale_method='rms')
print("inliers:", res['inliers'].sum())
print("relative distance error:", pairwise_distance_error(res['kpa_norm'], res['kpb_norm']))

# To create canonical images, use your existing warp_image_to_canonical with M_a and M_b.
```

```python
import numpy as np
import cv2

def compute_shared_canvas(M_a, M_b, img_a_shape, img_b_shape, scale=100.0, margin=20):
    """
    Compute a shared pixel canvas and full warp matrices for both images.

    Inputs:
      M_a, M_b : 2x3 (orig -> canonical) affine matrices
      img_x_shape : (H, W, ...) shapes for images
      scale : pixels per canonical unit
      margin: pixel margin around union bbox

    Returns:
      M_a_full, M_b_full : 2x3 warpAffine matrices mapping original image -> shared canvas pixels
      canvas_size : (W, H)
      S : 3x3 canonical->pixel homogeneous transform (for debugging)
    """
    def canonical_corners(M, h, w):
        corners = np.array([[0,0,1],[w,0,1],[w,h,1],[0,h,1]], dtype=np.float64)  # 4x3
        M_h = np.vstack([M, [0.0,0.0,1.0]])
        canon = (M_h @ corners.T).T[:, :2]   # 4x2 canonical coords
        return canon

    ha, wa = img_a_shape[:2]
    hb, wb = img_b_shape[:2]
    ca = canonical_corners(M_a, ha, wa)
    cb = canonical_corners(M_b, hb, wb)

    all_c = np.vstack([ca, cb])
    min_xy = all_c.min(axis=0)
    max_xy = all_c.max(axis=0)

    size_xy = (max_xy - min_xy) * scale
    out_w = int(np.ceil(size_xy[0] + 2 * margin))
    out_h = int(np.ceil(size_xy[1] + 2 * margin))

    tx = -min_xy[0] * scale + margin
    ty = -min_xy[1] * scale + margin
    S = np.array([[scale, 0.0, tx],
                  [0.0, scale, ty],
                  [0.0, 0.0, 1.0]], dtype=np.float64)

    def full(M):
        M_h = np.vstack([M, [0.0,0.0,1.0]])
        M_full = (S @ M_h)[:2,:].astype(np.float32)
        return M_full

    return full(M_a), full(M_b), (out_w, out_h), S

def warp_both_to_canvas(img_a, img_b, M_a_full, M_b_full, canvas_size):
    w, h = canvas_size
    # ensure 3-channel output for blending
    def to3(img):
        if img is None:
            raise FileNotFoundError("image is None")
        if img.ndim == 2:
            return cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
        return img.copy()
    A3 = to3(img_a)
    B3 = to3(img_b)
    wa = cv2.warpAffine(A3, M_a_full, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT, borderValue=0)
    wb = cv2.warpAffine(B3, M_b_full, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT, borderValue=0)
    return wa, wb

def draw_matches_overlay(wa, wb, pts_a_orig, pts_b_orig, M_a_full, M_b_full, idx_a=None, idx_b=None, inlier_mask=None, alpha=0.5):
    """
    - wa, wb: warped images on the same canvas (pixel)
    - pts_a_orig, pts_b_orig: original keypoints arrays (Nx2) corresponding to mapping order
    - M_*_full: full warp matrices (orig -> canvas pixels)
    - idx_a, idx_b: indices of the matched subset (if pts were extracted); if None assume 1-to-1 order
    - inlier_mask: boolean mask for matches (same length as pts arrays)
    """
    h, w = wa.shape[:2]
    blend = cv2.addWeighted(wa, alpha, wb, 1-alpha, 0)

    # map points to pixel coords on the canvas
    def map_pts(pts, M_full):
        pts_h = np.hstack([np.asarray(pts, dtype=np.float64), np.ones((len(pts),1))])
        mapped = (M_full @ pts_h.T).T[:, :2]
        return mapped.round().astype(int)

    if idx_a is None:
        pa = pts_a_orig
        pb = pts_b_orig
    else:
        pa = pts_a_orig[idx_a]
        pb = pts_b_orig[idx_b]

    pa_px = map_pts(pa, M_a_full)
    pb_px = map_pts(pb, M_b_full)

    # draw matches: green for inliers, red for others
    vis = blend.copy()
    for i, (p, q) in enumerate(zip(pa_px, pb_px)):
        if inlier_mask is None or inlier_mask[i]:
            color = (0, 255, 0)
            thickness = 1
        else:
            color = (0, 0, 255)
            thickness = 1
        cv2.circle(vis, tuple(p.tolist()), 3, color, -1)
        cv2.circle(vis, tuple(q.tolist()), 3, color, -1)
        cv2.line(vis, tuple(p.tolist()), tuple(q.tolist()), color, thickness)
    return vis, pa_px, pb_px

def make_side_by_side(wa, wb, scale_to=None):
    # place images side-by-side (same height)
    ha, wa_w = wa.shape[:2]
    hb, wb_w = wb.shape[:2]
    H = max(ha, hb)
    def pad(img, H):
        h = img.shape[0]
        if h==H:
            return img
        pad = np.zeros((H - h, img.shape[1], 3), dtype=img.dtype)
        return np.vstack([img, pad])
    A_p = pad(wa, H)
    B_p = pad(wb, H)
    return np.hstack([A_p, B_p])
```

```python
# assume you ran: res = normalize_pairs_consistent(...)
pts = data['matches']
kps_a = data['keypoints0']
kps_b = data['keypoints1']

# use only matched arrays (ordered)
pts_a_all, pts_b_all, idx_a_all, idx_b_all = build_pairwise(pts, kps_a, kps_b)

# optionally use RANSAC inliers you already computed
inlier_mask = res['inliers']            # boolean mask for the matched subset
M_a = res['M_a']                        # orig->canonical (2x3)
M_b = res['M_b']

# build shared canvas and full warps
M_a_full, M_b_full, canvas_size, S = compute_shared_canvas(M_a, M_b, image_a.shape, image_b.shape, scale=100.0, margin=20)

# warp images
wa, wb = warp_both_to_canvas(image_a, image_b, M_a_full, M_b_full, canvas_size)

# overlay visualization
overlay, pa_px, pb_px = draw_matches_overlay(
    wa, wb,
    pts_a_orig=kps_a, pts_b_orig=kps_b,      # original full keypoint arrays
    M_a_full=M_a_full, M_b_full=M_b_full,
    idx_a=idx_a_all, idx_b=idx_b_all,
    inlier_mask=inlier_mask
)
cv2.imwrite("overlay_matches.jpg", overlay)
cv2.imwrite("canonical_side_by_side.jpg", make_side_by_side(wa, wb))
```
